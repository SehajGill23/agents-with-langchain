# Text Prompts

Text prompts are strings - Ideal for straight-forward generation tasks where you don't need to retrain conversation history, example just using it like this: model.invoke("What is the capital of France?")

```python
import os
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model 
load_dotenv()  # Load environment variables from .env file

os.environ["OPENAI_API_KEY"]=os.getenv("OPENAI_API_KEY")

model = init_chat_model("gpt-4.1") # initialize a chat model with the name "gpt-4"
response = model.invoke("What is the capital of France?")
response # the output is of type AIMessage, which is a class that represents a message from the AI model. It has a text attribute that contains the text of the message.
```

# Message Prompts

In LangChain, a message represents one entry in a conversation between the user, model, system, or tool.

Common message types:

- SystemMessage  → instructions for the model
- HumanMessage   → user input
- AIMessage      → model response
- ToolMessage    → result returned by a tool

Each Message type has attributes -> role, content, type, name, id 

Alternativly, we can pass in a list of messages to the model by providing a list of message objects.

```python
from langchain.messages import AIMessage, HumanMessage, SystemMessage

messages = [
    # SystemMessage is a instruction message that provides context to the model. 
    #It has a content attribute that contains the text of the message.

    SystemMessage(role="system", content="You are a helpful assistant."), 

    # HumanMessage is a message from the user that contains a question or request.
    HumanMessage(role="user", content="What is the capital of France?"),
]
response = model.invoke(messages) # invoke the model with a list of messages

 # the output is of type AIMessage, which is a class that represents a message from the AI model. It has a text attribute that contains the text of the message.
response.content
```

```python
from langchain.messages import AIMessage, HumanMessage, SystemMessage

messages = [
    # SystemMessage is a instruction message that provides context to the model. 
    #It has a content attribute that contains the text of the message.

    SystemMessage(role="system", content="You are a helpful math assistant."), 
    # HumanMessage is a message from the user that contains a question or request.
    HumanMessage(role="user", content="What is 2 +2 ?"),

    AIMessage(role="assistant", content="The answer is 4."), # AIMessage is a message from the model that contains a response to the user's question or request.

    HumanMessage(role="user", content="What if we change it from + to * in the prev question what will be the answer?"),
]
response = model.invoke(messages) # invoke the model with a list of messages

 # the output is of type AIMessage, which is a class that represents a message from the AI model. It has a text attribute that contains the text of the message.
print(response.content)
print(f"Usage metadata: {response.usage_metadata}") 
```

```python
# Here we are simulating a tool call in the middle of a conversation, by hard coding. 
from langchain.messages import AIMessage, HumanMessage, SystemMessage, ToolMessage

ai_message = AIMessage(
    role="assistant", content=[],
    tool_calls=[{"name": "get_weather", "args": {"city": "San Francisco"}, "id": "tool_call_001"}]
    )

tool_message = ToolMessage(
    role="tool", content="The weather in San Francisco is sunny and 75 degrees.",
    tool_name="get_weather", tool_args={"city": "San Francisco"}, tool_call_id="tool_call_001"
    )

messages = [
    SystemMessage(role="system", content="You are a helpful Weather agent."),
    HumanMessage(role="user", content="What is the weather in San Francisco?"),
    ai_message,
    tool_message
]

response = model.invoke(messages) # invoke the model with a list of messages
print(response.content)
```
