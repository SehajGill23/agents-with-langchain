# Streaning the output of Model

```python
import os
from dotenv import load_dotenv

# ⚠️ init_chat_model creates a language model. create_agent creates a complete agent that can use a language model and tools.
from langchain.chat_models import init_chat_model 
load_dotenv()  # Load environment variables from .env file

os.environ["OPENAI_API_KEY"]=os.getenv("OPENAI_API_KEY")

model = init_chat_model("gpt-4.1") # initialize a chat model with the name "gpt-4"
response = model.stream("Why do parrots talk?")

for chunk in response:
    print(chunk.text, end="", flush=True)  # Print each chunk of the response as it is received, end with an empty string to avoid newlines, and flush the output buffer to ensure immediate display
```

# Batch

Batching is a collection of independent requests to a model can significantly improve performance and reduce costs, as the processing can be done in parallel.


```python
responses = model.batch([
    "why do parrots talk?",
    "why do dogs bark?",
    "why do cats meow?"
], config={"max_concurrency": 3}) 

for chunk in responses:
    print(chunk.text, end="", flush=True)  # Print each chunk of the response as it is received, end with an empty string to avoid newlines, and flush the output buffer to ensure immediate display
```
