# Model Inetegration with OpenAI, Gemini and Groq

```python
from langchain.chat_models import init_chat_model # langchain.chat_models is a module that contains the init_chat_model function which is used to initialize a chat model
model = init_chat_model("gpt-4.1") # initialize a chat model with the name "gpt-4"
model
```

```python
response = model.invoke("Hello how are you?")
response.content
```

## Alternate way is to use ChatOpenAI

```python
#Alternate way of using openai is chatOpenAI

from langchain_openai import ChatOpenAI

model = ChatOpenAI(model_name="gpt-4.1") # initialize a chat model with the name "gpt-4"
response = model.invoke("Hello how are you?")
response.content
```
